```python
def validate_yes_no(exit_choice):
    exit_choice = exit_choice.strip().upper()
    if exit_choice == 'YES':
        return True
    elif exit_choice == "NO":
        return False
    else:
        return None
```

```python
validate_yes_no('yes')
```

```python
validate_yes_no('no')
validate_yes_no('yes')
validate_yes_no('yEs')
validate_yes_no('    yes')
validate_yes_no('    no')
validate_yes_no('    nO')
```
